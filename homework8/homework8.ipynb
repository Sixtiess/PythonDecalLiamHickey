```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    # y = A * sin(w * x), A is the amplitude and w is the frequency
    return A * np.sin(w * x)
```

```python
x = np.linspace(0, 2 * np.pi, 1000)

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(25, 5))
linestyles = ["-", "--", "-.", ":", (0, (5, 1))]

# First subplot: change only the amplitude (frequency stays at 1)
for A, ls in zip(amplitude, linestyles):
    y = make_sine_wave(x, A, 1)
    axes[0].plot(x, y, linestyle=ls, label=f"A = {A}, w = 1")
axes[0].set_title("Changing the Amplitude", fontsize=16)
axes[0].set_xlabel("x (radians)", fontsize=14)
axes[0].set_ylabel("y = A sin(wx)", fontsize=14)
axes[0].legend(fontsize=11)

# Second subplot: change only the frequency (amplitude stays at 1)
for w, ls in zip(frequency, linestyles):
    y = make_sine_wave(x, 1, w)
    axes[1].plot(x, y, linestyle=ls, label=f"A = 1, w = {w}")
axes[1].set_title("Changing the Frequency", fontsize=16)
axes[1].set_xlabel("x (radians)", fontsize=14)
axes[1].set_ylabel("y = A sin(wx)", fontsize=14)
axes[1].legend(fontsize=11, loc="upper right")

# Third subplot: change the amplitude and the frequency together
for A, w, ls in zip(amplitude, frequency, linestyles):
    y = make_sine_wave(x, A, w)
    axes[2].plot(x, y, linestyle=ls, label=f"A = {A}, w = {w}")
axes[2].set_title("Changing the Amplitude and Frequency Together", fontsize=16)
axes[2].set_xlabel("x (radians)", fontsize=14)
axes[2].set_ylabel("y = A sin(wx)", fontsize=14)
axes[2].legend(fontsize=11, loc="upper right")

plt.show()
```

What each plot shows:

Amplitude only: every wave crosses zero at the same spots, but the peaks get taller as A goes up. So the amplitude just stretches the wave up and down.

Frequency only: the peaks are all the same height (1), but more waves fit in 0 to 2pi as w goes up. w = 5 completes 5 full cycles in that range.

Both together: the waves get taller and faster at the same time, so the high numbers look like a tall, squished wave and the low numbers look like a short, stretched one.

For the bonus I gave each line a different linestyle and put the amplitude and frequency in the legend labels.
