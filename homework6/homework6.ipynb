## 3.1 1D NumPy Arrays

```python
lst = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(lst)

print("Times 4:", arr * 4)
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard deviation:", np.std(arr))

min_idx = np.argmin(arr)
max_idx = np.argmax(arr)
print("Index of smallest:", min_idx, "-> value", arr[min_idx])
print("Index of largest:", max_idx, "-> value", arr[max_idx])

print("np.min:", np.min(arr), "| np.max:", np.max(arr))
print("Match argmin/argmax values?", arr[min_idx] == np.min(arr) and arr[max_idx] == np.max(arr))

print("Values > 20:", arr[arr > 20])
```

## 3.2 2D NumPy Arrays

```python
matrix = np.array([
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
])

print("Shape:", matrix.shape)
print("Rows:", matrix.shape[0], "| Columns:", matrix.shape[1])
print("Third column:", matrix[:, 2])
print("Second row:", matrix[1])
print("Times 2:")
print(matrix * 2)
print("Mean of each column:", matrix.mean(axis=0))
print("Overall mean:", matrix.mean())
print("Max of each row:", matrix.max(axis=1))
print("Min of each column:", matrix.min(axis=0))
```

```python
flat = matrix.flatten()
print("Flattened:", flat)
print("New shape:", flat.shape)
# flatten() collapses the 2D array (3x4) into a 1D copy of length 12, reading row by row.

flat_min = np.argmin(matrix)
flat_max = np.argmax(matrix)
print("Flat index of smallest:", flat_min, "| largest:", flat_max)

print("Smallest at (row, col):", tuple(int(i) for i in np.unravel_index(flat_min, matrix.shape)))
print("Largest at (row, col):", tuple(int(i) for i in np.unravel_index(flat_max, matrix.shape)))

print("Values > 15:", matrix[matrix > 15])
print("Each row sorted:")
print(np.sort(matrix, axis=1))
```

## 4 Prime Star Systems

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

def prime_rows(arr):
    """Return only the rows that contain at least one prime number."""
    prime_mask = np.vectorize(is_prime)(arr)
    return arr[prime_mask.any(axis=1)]

stars_mag = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])
print(prime_rows(stars_mag))
```

## 5 Let's Play Checkers!

```python
def checkerboard():
    board = np.zeros((8, 8), dtype=int)   # 5.1
    board[0::2, 0::2] = 1                 # 5.2 odd-numbered rows (1st, 3rd, ...) start with 1
    board[1::2, 1::2] = 1                 # 5.3 even-numbered rows start with 0
    return board

def reverse_checkerboard():
    return 1 - checkerboard()             # 5.4 same board, starting with 0

print(checkerboard())
print(reverse_checkerboard())
```

## 6 The Expanding Universe

```python
def expansion(words, n):
    """Put n spaces between each letter of every string in the array."""
    spacer = " " * n
    return np.array([spacer.join(w) for w in words])

universe = np.array(['galaxy', 'clusters'])
print(repr(expansion(universe, 1)))
print(repr(expansion(universe, 2)))
```

## 7 Second-Brightest Star

```python
def secondDimmest(stars):
    """Second-largest value in each row (matches the sample output)."""
    return np.sort(stars, axis=1)[:, -2]

np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
print(stars)
print(secondDimmest(stars))

# Check against the example from the assignment
example = np.array([[1123, 1456, 1789, 1324, 1876],
                    [1567, 1987, 1678, 1405, 1589],
                    [1345, 1654, 1523, 1109, 1923],
                    [1298, 1890, 1367, 1784, 1432],
                    [1823, 1756, 1489, 1672, 1550]])
print(secondDimmest(example))
```
