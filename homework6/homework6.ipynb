## 3.1 1D NumPy Arrays

```python
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list) # make the list into a numpy array so we can do math on it

times_four = arr * 4 # multiplying the array multiplies every number in it by 4
print(times_four)

mean = np.mean(arr)
print(mean)

median = np.median(arr)
print(median)

std = np.std(arr)
print(std)

# argmin and argmax tell us WHERE the smallest and largest numbers are (their index), not what they are
smallest_index = np.argmin(arr)
largest_index = np.argmax(arr)
print(smallest_index)
print(largest_index)

# now use those spots to grab the real numbers out of arr
print(arr[smallest_index])
print(arr[largest_index])

# np.min and np.max just hand us the numbers themselves
print(np.min(arr))
print(np.max(arr))
# yep, these match what we got using the indexes

# arr > 20 makes a list of True/False for each number, and using that as the index keeps only the Trues
greater_than_20 = arr[arr > 20]
print(greater_than_20)
```

## 3.2 2D NumPy Arrays

```python
matrix = np.array([
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
])

print(matrix.shape) # gives (rows, columns)
# so 3 rows and 4 columns

print(matrix[:, 2]) # third column: take every row (:) and column number 2 (counting starts at 0)
print(matrix[1]) # second row is row number 1 since we start counting at 0

print(matrix * 2)

# axis=0 means go down each column, axis=1 means go across each row
column_means = np.mean(matrix, axis=0)
print(column_means)

overall_mean = np.mean(matrix)
print(overall_mean)

row_maxes = np.max(matrix, axis=1) # biggest number in each row
print(row_maxes)

column_mins = np.min(matrix, axis=0) # smallest number in each column
print(column_mins)
```

```python
flat = matrix.flatten()
print(flat)
print(flat.shape)
# flatten squished the 3x4 grid into one long line of 12 numbers, reading it row by row

flat_min_index = np.argmin(matrix) # with no axis, it counts spots as if the grid was one long line
flat_max_index = np.argmax(matrix)
print(flat_min_index)
print(flat_max_index)

# unravel_index turns that spot in the long line back into (row, column) in the grid
min_position = np.unravel_index(flat_min_index, matrix.shape)
max_position = np.unravel_index(flat_max_index, matrix.shape)
print(min_position)
print(max_position)

greater_than_15 = matrix[matrix > 15]
print(greater_than_15)

sorted_rows = np.sort(matrix, axis=1) # axis=1 sorts each row by itself
print(sorted_rows)
```

## 4 Prime Star Systems

```python
# helper function that checks if one number is prime
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n): # try dividing by every number from 2 up to n-1
        if n % i == 0:
            return False # something divides evenly, so it isn't prime
    return True

def prime_rows(arr):
    rows_with_prime = [] # the rows we're keeping
    for row in arr:
        has_prime = False
        for number in row:
            if is_prime(number):
                has_prime = True
        if has_prime:
            rows_with_prime.append(row)
    return np.array(rows_with_prime)

stars_mag = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])
print(prime_rows(stars_mag))
```

## 5 Let's Play Checkers!

```python
def checkerboard():
    board = np.zeros((8, 8), dtype=int) # start with an 8x8 board full of zeros

    # [start::2] means every other spot, starting from "start"
    board[0::2, 0::2] = 1 # 1st, 3rd, 5th, 7th rows get a 1 in every other column starting at the first
    board[1::2, 1::2] = 1 # 2nd, 4th, 6th, 8th rows get a 1 in every other column starting at the second
    return board

def reverse_checkerboard():
    board = checkerboard()
    board = 1 - board # this flips every 1 to a 0 and every 0 to a 1
    return board

print(checkerboard())
print(reverse_checkerboard())
```

## 6 The Expanding Universe

```python
def expansion(words, n):
    spaces = " " * n # the spaces to put between the letters
    expanded = []
    for word in words:
        expanded.append(spaces.join(word)) # puts the spaces in between every letter
    return np.array(expanded)

universe = np.array(['galaxy', 'clusters'])
print(expansion(universe, 1))
print(expansion(universe, 2))
```

## 7 Second-Brightest Star

```python
# the example output in the assignment gives the second LARGEST number in each row, so that's what this does
def secondDimmest(stars):
    sorted_stars = np.sort(stars, axis=1) # sort each row from smallest to largest
    return sorted_stars[:, -2] # -2 means second from the end, so the second biggest

np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
print(stars)
print(secondDimmest(stars))

# same thing again, but with the array the assignment prints so we can compare
example = np.array([[1123, 1456, 1789, 1324, 1876],
                    [1567, 1987, 1678, 1405, 1589],
                    [1345, 1654, 1523, 1109, 1923],
                    [1298, 1890, 1367, 1784, 1432],
                    [1823, 1756, 1489, 1672, 1550]])
print(secondDimmest(example))
```
